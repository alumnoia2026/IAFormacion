```python
df = pd.DataFrame({
    "nombre": ["Ana", "Bea", "Carlos", "Diego", "Elena"],
    "departamento": ["IT", "Ventas", "IT", "RRHH", "Ventas"],
    "salario": [32000, 28000, 35000, 30000, 27000]
})

# 1. Valores únicos
print(df["departamento"].unique())

# 2. Número de empleados por departamento
print(df["departamento"].value_counts())

# 3. Porcentaje de cada departamento
print(df["departamento"].value_counts(normalize=True) * 100)
```

```python
ventas = pd.DataFrame({
    "producto": ["A", "B", "C", "D"],
    "cantidad": [10, None, 7, 15],
    "precio": [5.0, 3.5, None, 8.0]
})

# 1. Contar valores nulos por columna
print(ventas.isnull().sum())

# 2. Rellenar los nulos de cantidad con la mediana
ventas["cantidad"] = ventas["cantidad"].fillna(ventas["cantidad"].median())

# 3. Rellenar los nulos de precio con 0
ventas["precio"] = ventas["precio"].fillna(0)

# 4. Comprobar que no quedan valores nulos
print(ventas.isnull().sum())

print(ventas)
```

```python
pedidos = pd.DataFrame({
    "cliente": ["Ana", "Bea", "Ana", "Carlos"],
    "importe": [50, 30, 50, 20]
})

# 1. Comprobar cuántas filas están duplicadas
print(pedidos.duplicated().sum())

# 2. Eliminar las filas duplicadas
pedidos = pedidos.drop_duplicates()

# 3. Comprobar la nueva forma del DataFrame
print(pedidos.shape)

print(pedidos)
```

```python
# 1. Crear la columna categoria
df["categoria"] = df["salario"].apply(
    lambda salario: "Alto" if salario >= 30000 else "Bajo"
)

print(df)

# 2. Contar empleados por categoría
print(df["categoria"].value_counts())

# 3. Filtrar empleados de categoría Alto
altos = df[df["categoria"] == "Alto"]

print(altos)
```

```python
# 1. Salario medio por departamento
media = df.groupby("departamento")["salario"].mean()

print(media)

# 2. Media y máximo por departamento
resumen = df.groupby("departamento")["salario"].agg(
    ["mean", "max"]
)

print(resumen)

# 3. Ordenar por salario medio descendente
resumen = resumen.sort_values("mean", ascending=False)

print(resumen)
```
