```python
# Datos
datos = np.random.normal(5, 1.5, 300)
x_d = np.random.rand(50)
y_d = x_d + np.random.rand(50) * 0.5

# Crear una figura con 1 fila y 3 columnas
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# 1. Histograma
axes[0].hist(datos)
axes[0].set_title("Histograma")

# 2. Gráfico de dispersión
axes[1].scatter(x_d, y_d)
axes[1].set_title("Dispersión")
axes[1].set_xlabel("x_d")
axes[1].set_ylabel("y_d")

# 3. Gráfico de barras
categorias = ["A", "B", "C"]
valores = [23, 45, 31]

axes[2].bar(categorias, valores)
axes[2].set_title("Gráfico de barras")
axes[2].set_xlabel("Categorías")
axes[2].set_ylabel("Valores")

# Ajustar el espacio entre gráficas
plt.tight_layout()

# Mostrar la figura
plt.show()
```

```python
# Cargar datos
tips = sns.load_dataset("tips")

# Explorar
print(tips.columns)
print(tips.shape)
print(tips.head())

# Cambiar la paleta
sns.set_palette("Set2")

# Boxplot
sns.boxplot(data=tips, x="day", y="total_bill")

plt.title("Cuenta total por día")
plt.show()
```

```python
# Cargar el dataset
tips = sns.load_dataset("tips")

# Histograma con curva de densidad
sns.histplot(
    tips["total_bill"],
    bins=20,
    kde=True,
    color="#059669"
)

# Título
plt.title("Distribución de la cuenta total")

# Mostrar gráfico
plt.show()
```

```python
# Cargar el dataset
tips = sns.load_dataset("tips")

# 1. Calcular la matriz de correlación
corr = tips.corr(numeric_only=True)

# Mostrar la matriz por pantalla
print(corr)

# 2 y 3. Crear el mapa de calor con los valores
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

# Título
plt.title("Mapa de calor de correlaciones")

# Mostrar gráfico
plt.show()
```
