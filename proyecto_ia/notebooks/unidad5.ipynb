```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
ventas = [120, 145, 130, 170, 160, 190]

# 1. Crear figura y ejes
fig, ax = plt.subplots(figsize=(8, 4))

# 2. Dibujar las ventas
ax.plot(meses, ventas)

# 3. Añadir título
ax.set_title("Ventas del primer semestre")

# 4. Etiquetar los ejes
ax.set_xlabel("Mes")
ax.set_ylabel("Ventas (€)")

# 5. Mostrar la gráfica
plt.show()
```

```python

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
tienda_a = [120, 145, 130, 170, 160, 190]
tienda_b = [100, 110, 135, 140, 175, 165]

# Crear figura y ejes
fig, ax = plt.subplots(figsize=(8, 4))

# Dibujar las dos líneas
ax.plot(meses, tienda_a, label="Tienda A", color="#0891B2")
ax.plot(meses, tienda_b, label="Tienda B", color="#059669")

# Título y etiquetas
ax.set_title("Comparación de ventas")
ax.set_xlabel("Mes")
ax.set_ylabel("Ventas (€)")

# Mostrar leyenda
ax.legend()

# Mostrar gráfica
plt.show()
```

```python

cursos = ["Python", "IA", "Datos", "Web"]
alumnos = [30, 45, 25, 38]

# Colores para cada barra
colores = ["#0891B2", "#059669", "#F59E0B", "#EF4444"]

# Crear figura y ejes
fig, ax = plt.subplots(figsize=(8, 4))

# 1 y 2. Gráfico de barras verticales
ax.bar(cursos, alumnos, color=colores)

# 3. Añadir título
ax.set_title("Alumnos por curso")

# Etiquetas
ax.set_xlabel("Curso")
ax.set_ylabel("Número de alumnos")

# Mostrar gráfica
plt.show()
```

```python

datos = np.random.normal(loc=170, scale=10, size=500)

# Dibujar el histograma
plt.hist(datos, bins=20, edgecolor="white")

# Añadir título
plt.title("Distribución de estaturas")

# Etiqueta del eje X
plt.xlabel("Altura (cm)")

# Etiqueta del eje Y
plt.ylabel("Frecuencia")

# Mostrar el gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt

# Datos
horas = [1, 2, 2, 3, 4, 5, 5, 6, 7, 8]
notas = [3, 4, 5, 5, 6, 7, 8, 8, 9, 10]

# Crear gráfico de dispersión
plt.scatter(horas, notas, alpha=0.7, s=80)

# Título y etiquetas
plt.title("Relación entre horas de estudio y nota")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota obtenida")

# Ajustar márgenes
plt.tight_layout()

# Guardar la gráfica como PNG
plt.savefig("mi_grafica.png", dpi=150, bbox_inches="tight")

# Mostrar la gráfica
plt.show()
```

```python

datos = np.random.normal(5, 1.5, 300)

x_d = np.random.rand(50)
y_d = x_d + np.random.rand(50) * 0.5

# Crear una figura con 1 fila y 3 columnas
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Gráfica 1: Histograma
axes[0].hist(datos, bins=20, edgecolor="white")
axes[0].set_title("Histograma")

# Gráfica 2: Gráfico de dispersión
axes[1].scatter(x_d, y_d)
axes[1].set_title("Gráfico de dispersión")

# Gráfica 3: Gráfico de barras
categorias = ["A", "B", "C"]
valores = [23, 45, 31]

axes[2].bar(categorias, valores)
axes[2].set_title("Gráfico de barras")

# Ajustar el diseño
plt.tight_layout()

# Mostrar la figura
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar dataset
tips = sns.load_dataset("tips")

# Calcular la correlación
corr = tips.corr(numeric_only=True)

# Mostrar la matriz
print(corr)

sns.heatmap(corr)

plt.title("Mapa de calor de correlación")

plt.show()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Mapa de calor de correlación")

plt.show()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Mapa de calor de correlación")

plt.show()

print(corr)
```
