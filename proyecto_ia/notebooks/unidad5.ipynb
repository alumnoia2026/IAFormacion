```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]
ventas = [120, 145, 130, 170, 160, 190]
fig, ax =plt.subplots(figsize=(8, 4))
ax.plot(meses, ventas, marker='o', linestyle='-', color='b')
ax.set_title("Ventas del primer semestre")
ax.set_xlabel("Meses")
ax.set_ylabel("Ventas(€)")
plt.show()
```

```python
tienda_a = [120, 145, 130, 170, 160, 190]
tienda_b = [100, 110, 135, 140, 175, 165]
fig, ax =plt.subplots(figsize=(8, 4))
ax.plot(meses, tienda_a, marker='o', linestyle='-', color='#0891B2', label='Tienda A')
ax.plot(meses, tienda_b, marker='o', linestyle='-', color='#059669', label='Tienda B')
ax.legend()
```
